```python
imonte=10

ifilter = 'concat'
all_models = glob(f"../models/{ifilter}/cdan_baha2dark_pre_squeezenet1_aw_1_pad_shear_avgpool_gauss_seed_1_nob1_ft_tweaked_align_*_best.pth")


#output_name = f"pickles/alignment_test_results.pkl"

#if os.path.isfile( output_name ):
#    all_results = pkl.load(open(output_name,"rb"))
#else:
all_results = {}

for imodel in tqdm(all_models):
   
    name = imodel.split('_')[-2]
    print(name)
    args.jwst_filter = 'concat'
    args.apply_intrinsic_ell = 1.
    zs = {
    'f115w':1.36,
    'f150w':1.56,
    'concat':1.46
    }
    domain = {
    'tgt':'darkskies_obs',
    'src':'bahamas_obs'
    }
    args.ignore_dataset = [''] # Althopiugh i ignored during training i want to see duringn testing.
    all_results[name]  = {'src':[],'tgt':[]}

    args.log_mass_cut=0. # Make sure i test on equal mass clusters

    args.unbalance = True
    for i in range(imonte):
        args.zs = zs['concat']

        for idomain in domain.keys():

            target_domain = domain[idomain]
            results = get_probabilities( 
                    target_domain,
                    [imodel],
                    args,
                    quiet=True
            )
            del results['data_loaders']

            all_results[name][idomain].append( results )

#pkl.dump(all_results, open(output_name,"wb"))
```

```python
all_results['No Adaptation'] = all_results['No Adpatation'].copy()
```

```python
fig, axarr = plt.subplots(1,2,figsize=(7,4))
cdm = 1e-3
#all_results = pkl.load(open(f"pickles/alignment_test_results.pkl",'rb'))
color={'src':'r','tgt':'b'}
domain = {
    'tgt':'darkskies_obs',
    'src':'bahamas_obs'
    }
for imx, imodel in enumerate(all_results.keys()):
    ax = axarr[imx]
    ref = None
    for itgt, target in enumerate(['src','tgt']):
        tgt = get_threshold_for_cross( 
                all_results[imodel][target], 
                mass_cut=None, 
                dataset=domain[target].split('_')[0],
                quiet=False)





        tgt['cross_sections'][0] += cdm
        ax.errorbar( tgt['cross_sections'], 
                   1-tgt['thresholds'],
                    tgt['threshold_err'].T/np.sqrt(30),
                    fmt='-v', label=f"{target.upper()} ({domain[target].split('_')[0]})", 
                    capsize=2,
                    markersize=8, color=color[target])
        ax.set_xlabel("Self-interaction cross-section [cm$^2$/g]")
        ax.legend()
    #ax.set_xscale('log')

    ax.set_xlabel("Self-interaction cross-section [cm$^2$/g]")
    #ax.set_ylabel("Model threshold output")

    #ax.set_xscale('log')
    ax.text(0.05,0.85,f"{imodel}", transform=ax.transAxes,fontsize=12)
    ax.set_xscale('log')
```

# Unbalanced vs balance

```python
unbalanced = pkl.load(open(f"pickles/all_models_concat_ub_results.pkl","rb"))
balanced = pkl.load(open(f"pickles/all_models_concat_results.pkl","rb"))

fig, axarr = plt.subplots(1,2, figsize=(6,3))

for ibx, balance in enumerate([ unbalanced, balanced]):
    ax = axarr[ibx]
    for itgt, target in enumerate(['src','tgt']):
            all_thresholds = []
            for imodel in unbalanced.keys():

                seed = float(imodel.split('_')[1])

                tgt = get_threshold_for_cross( 
                    balance[imodel][target], 
                    mass_cut=[14.7,15], 
                    integrated_mass=True,
                    dataset=domain[target].split('_')[0],
                    quiet=False)


                all_thresholds.append(tgt['thresholds'])

            all_thresholds = np.array(all_thresholds)
            means = np.nanmean(all_thresholds,axis=0)
            print(means)
            errors = np.std(all_thresholds,axis=0) / np.sqrt(all_thresholds.shape[0]) 
            cross = tgt['cross_sections']
            cross[ cross == 0] = 3e-3
            ax.errorbar( cross, 
                   1-means,
                    errors,
                    fmt='-o', 
                    capsize=2,
                    markersize=8, color=color[target])
            
    ax.set_xscale('log')
```
